```python
from langchain_ollama import ChatOllama
model = ChatOllama(model="llama3.2:latest")
print('Model is ready')
```

```python
# Tools definition
from langchain_community.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get weather for a city"""
    return f"The weather in the city {city} is Sunny"

@tool
def get_stock_price(stock: str) -> str:
    """Returns the stock price for a given stock"""
    return f"The stock price of {stock} is 25.50"

get_weather
get_stock_price
print('Tools are ready')
```

```python
# Creating an agent
from langchain.agents import create_agent

agent = create_agent(
    model=model,
    tools=[get_weather, get_stock_price],
    system_prompt="You are a helpful assistant."
)
print('AI agent is ready')
```

```python
response = agent.invoke({
    "messages": [{
        "role": "user",
        "content": "Which is the capital city of India?"
    }]
})

response
```

```python
response = agent.invoke({
    "messages": [{
        "role": "user",
        "content": "How is the weather in Pune city today?"
    }]
})

response
```
